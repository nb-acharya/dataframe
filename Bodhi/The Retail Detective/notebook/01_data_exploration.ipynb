```python
df.groupby("Region")[["Sales", "Profit"]].sum().sort_values("Profit", ascending=False)
```

```python
df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Quantity=("Quantity", "sum")
).sort_values("Profit", ascending=False)
```

```python
df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).assign(
    Profit_Margin=lambda x: x["Profit"] / x["Sales"]
).sort_values("Profit_Margin", ascending=False)
```
